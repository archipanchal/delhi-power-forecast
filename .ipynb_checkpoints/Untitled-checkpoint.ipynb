```python
import pandas as pd
```

```python

# Load data from Excel file
file_path = r"C:\Users\Sakshi Shah\Downloads\archive\powerdemand_5min_2021_to_2024_with weather.csv"

```

```python

df = pd.read_csv(file_path)
```

```python
print(df)
```

```python
# Drop unnecessary columns if they exist
columns_to_drop = ["Unnamed: 0", "moving_avg_3"]
df = df.drop(columns=[col for col in columns_to_drop if col in df.columns], errors="ignore")

# Convert datetime column to proper format
df["datetime"] = pd.to_datetime(df["datetime"], errors='coerce')

# Set datetime as index
df.set_index("datetime", inplace=True)

# Handle missing values (forward fill)
df.fillna(method='ffill', inplace=True)

# Display summary
print(df.info())
print(df.head())
```

```python
# Create lag features (previous demand values)
df['lag_1'] = df['Power demand'].shift(1)
df['lag_3'] = df['Power demand'].shift(3)
df['lag_6'] = df['Power demand'].shift(6)

# Create rolling average features
df['rolling_mean_3'] = df['Power demand'].rolling(window=3).mean()
df['rolling_mean_6'] = df['Power demand'].rolling(window=6).mean()

# Extract time-based features
df['day_of_week'] = df.index.dayofweek  # Monday = 0, Sunday = 6
df['hour'] = df.index.hour  # Extract hour of the day
df['is_weekend'] = (df['day_of_week'] >= 5).astype(int)  # 1 for weekend, 0 for weekdays

# Handle missing values created by shifting
df.fillna(method='bfill', inplace=True)

# Display the updated DataFrame
print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Select only numeric columns for scaling
numeric_features = df.select_dtypes(include=[np.number]).columns

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Fit and transform the data
df_scaled = pd.DataFrame(scaler.fit_transform(df[numeric_features]), columns=numeric_features, index=df.index)

# Save the scaler for later use (to inverse transform predictions)
import joblib
joblib.dump(scaler, "scaler.pkl")

# Display scaled data
print(df_scaled.head())
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import joblib

# Select only numeric columns for scaling
numeric_features = df.select_dtypes(include=[np.number]).columns

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Fit and transform the data
df_scaled = pd.DataFrame(scaler.fit_transform(df[numeric_features]), columns=numeric_features, index=df.index)

# Save the scaler for later use (to inverse transform predictions)
joblib.dump(scaler, "scaler.pkl")

# Display scaled data
print(df_scaled.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python

# Define how many past time steps the LSTM should use
TIME_STEPS = 30  # Looking back 30 timesteps (~2.5 hours for 5-min data)
```

```python
# Function to create sequences for LSTM
def create_sequences(data, time_steps):
    X, Y = [], []
    for i in range(len(data) - time_steps):
        X.append(data.iloc[i : i + time_steps].values)  # Take past `time_steps` rows
        Y.append(data.iloc[i + time_steps]["Power demand"])  # Target is next step's demand
    return np.array(X), np.array(Y)
```

```python
# Create sequences
X, Y = create_sequences(df_scaled, TIME_STEPS)

# Split into train (80%) and test (20%) sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, shuffle=False)

# Print dataset shapes
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
print(f"Y_train shape: {Y_train.shape}, Y_test shape: {Y_test.shape}")
```
